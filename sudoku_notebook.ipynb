# Sudoku Game

- Step 1: Create the Board
- Step 2: Manually import number layout on the board to ensure it can be solved
- Step 3: Have the player input numbers into empty squares
- Step 4: Number placement - check where the number is placed and if it can be placed (needs to be empty)
- Step 5: Error check/Space check - in case the player puts a number in any row/square where it already exists - return boolean
- Step 6: Have the solution so we can reference to what the winning outcome is
- Step 7: Use the previous step to prevent player from entering wrong number in an empty square
- Step 8: Define the end of the game, when the board is full it means it is solved
- Step 9: If the player solved it, ask for a rematch
- BONUS: Have difficulty levels, or different board templates
