### Step 3: Have the player input numbers into empty squares

```python
# Create a function that allows the player to choose a number to place in the sudoku

def player_input(board):
    row_position = 0
    column_position = 0
    number = 0
    while row_position not in range(1,10):
        row_position = int(input("Choose row number between 1 and 9:"))
        if row_position > 9 or row_position < 1:
            print("Please choose a number between 1 and 9")

    while column_position not in range(1,10):
        column_position = int(input("Choose column number between 1 and 9:"))
        if column_position > 9 or column_position < 1:
            print("Please choose a number between 1 and 9")

    while number not in range(1,10):
        number = int(input("Place your number of choice:"))
        if number > 9 or number < 1:
            print("Please choose a number between 1 and 9")
    
    return  row_position, column_position, number
```

### Step 4: Number placement - check where the number is placed and if it can be placed (needs to be empty)

```python
# We want a function that places the player's input on the board

def number_placement(board, player_input):
    row, col, guess = player_input
    board[row][col] = str(guess)
    return board
```
